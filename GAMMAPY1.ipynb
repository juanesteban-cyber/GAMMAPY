```python
import astropy.units as u
from astropy.coordinates import SkyCoord
import matplotlib.pyplot as plt
```

```python
from gammapy.utils.check import check_tutorials_setup

# %matplotlib inline


check_tutorials_setup()
```

```python
from gammapy.maps import Map

gc_3fhl = Map.read("$GAMMAPY_DATA/fermi-3fhl-gc/fermi-3fhl-gc-counts.fits.gz")
```

```python
print(gc_3fhl)
```

```python
print(gc_3fhl.geom)
```

```python
print(f"Total number of counts in the image: {gc_3fhl.data.sum():.0f}")
```

```python
gc_3fhl.plot(stretch="sqrt")
plt.show()
```

```python
gc_3fhl_smoothed = gc_3fhl.smooth(kernel="gauss", width=0.1 * u.deg)
gc_3fhl_smoothed.plot(stretch="sqrt")
plt.show()
```

```python
# define center and size of the cutout region
center = SkyCoord(0, 0, unit="deg", frame="galactic")
gc_3fhl_cutout = gc_3fhl_smoothed.cutout(center, 9 * u.deg)
gc_3fhl_cutout.plot(stretch="sqrt")
plt.show()
```

```python
ax=gc_3fhl_cutout.plot(stretch="sqrt")


ax.set_autoscale_on(False)

# Add a rectangle with bottom left corner at pixel position (30, 50) with a
# width and height of 60 and 50 pixels respectively.
from matplotlib.patches import Circle
r = Circle((0,0),1, edgecolor='blue', facecolor='none', linewidth= 2, transform=ax.get_transform('galactic'))
ax.add_patch(r)

# Add three markers at (40, 30), (100, 130), and (130, 60). The facecolor is
# a transparent white (0.5 is the alpha value).
ax.scatter([40, 50, 130, 120], [30, 130, 60, 160], s=100, edgecolor='white', facecolor=(1, 1, 1, 0.5))
```
